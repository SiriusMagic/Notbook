```python
from yt_dlp import YoutubeDL
from pathlib import Path
from mutagen import File
import json

#Celda 3 — configurar carpetas

BASE_DIR = Path("/content/music")
BASE_DIR.mkdir(exist_ok=True)

DB_FILE = Path("/content/library.json")

library = []

#elda 4 — función de descarga

def download_audio(url, folder=BASE_DIR):
    opts = {
        "format": "bestaudio/best",
        "outtmpl": str(folder / "% (playlist_title|single)s/% (playlist_index|00)s - % (title)s.% (ext)s"),
        "ignoreerrors": True,
        "noplaylist": False,
        "postprocessors": [
            {
                "key": "FFmpegExtractAudio",
                "preferredcodec": "mp3",
                "preferredquality": "192",
            },
            {
                "key": "FFmpegMetadata",
            }
        ],
        "writethumbnail": True,
        "embedthumbnail": True,
    }

    with YoutubeDL(opts) as ydl:
        return ydl.extract_info(url, download=True)

#Celda 5 — pegar URL y descargar

url = input("Pega URL o playlist: ")
info = download_audio(url)
print("Descarga terminada")

#Celda 6 — escanear música descargada

def scan_music(folder=BASE_DIR):
    songs = []

    for file in folder.rglob("*.mp3"):
        audio = File(file, easy=True)

        title = audio.get("title", [file.stem])[0] if audio else file.stem
        artist = audio.get("artist", ["Desconocido"])[0] if audio else "Desconocido"
        album = audio.get("album", ["Desconocido"])[0] if audio else "Desconocido"

        songs.append({
            "title": title,
            "artist": artist,
            "album": album,
            "path": str(file),
            "mood": None
        })

    return songs

library = scan_music()
library[:5]

#Celda 7 — guardar biblioteca

with open(DB_FILE, "w", encoding="utf-8") as f:
    json.dump(library, f, ensure_ascii=False, indent=2)

print(f"Biblioteca guardada en {DB_FILE}")

#Celda 8 — clasificar sentimiento básico

MOOD_WORDS = {
    "amor": ["love", "amor", "contigo", "beso", "baby", "corazón", "romance"],
    "tristeza": ["sad", "triste", "llorar", "adiós", "alone", "solo", "perdí"],
    "fiesta": ["party", "dance", "baila", "perreo", "noche", "club"],
    "energia": ["gym", "power", "fuego", "duro", "rápido", "speed"],
    "calma": ["sleep", "relax", "calma", "piano", "lofi", "study"],
    "ira": ["hate", "odio", "rabia", "war", "enemy"]
}

def classify_mood(song):
    text = f"{song['title']} {song['artist']} {song['album']}".lower()

    scores = {}

    for mood, words in MOOD_WORDS.items():
        scores[mood] = sum(1 for word in words if word in text)

    best_mood = max(scores, key=scores.get)

    if scores[best_mood] == 0:
        return "desconocido"

    return best_mood

for song in library:
    song["mood"] = classify_mood(song)

library[:10]

#Celda 9 — ver canciones por sentimiento

def songs_by_mood(mood):
    return [song for song in library if song["mood"] == mood]

songs_by_mood("amor")

#Celda 10 — descargar carpeta como ZIP

!zip -r /content/music.zip /content/music

#Luego en Colab:

#from google.colab import files
#files.download("/content/music.zip")
```
