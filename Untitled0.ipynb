```python

def solicitar_url():
  """
  Solicita el URL de la música que desea descargar el usuario.

  Devuelve:
    URL de la música.
  """

  url = input("Ingrese el URL de la música que desea descargar: ")
  return url


!pip install pytube

import pytube
import urllib

# Solicitamos el URL de la música al usuario.
url = solicitar_url()

# Descargamos la canción en formato MP3.
youtube = pytube.YouTube(url)
stream = youtube.streams.get_audio_only()
stream.download()

# Descargamos la imagen de la canción.
thumbnail = youtube.thumbnail_url
urllib.request.urlretrieve(thumbnail, "thumbnail.jpg")

# Imprimimos el nombre del archivo de la canción descargada.
print(stream.default_filename)
```
